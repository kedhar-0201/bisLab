```python

import random
graph = [
    [0, 10, 0, 30, 100],
    [10, 0, 50, 0, 0],
    [0, 50, 0, 20, 10],
    [30, 0, 20, 0, 60],
    [100, 0, 10, 60, 0]
]
N = 5
Pa = 0.25
T = 100
source = 0
destination = 4
def fitness(path):
    cost = 0
    for i in range(len(path) - 1):
        u = path[i]
        v = path[i + 1]
        if graph[u][v] == 0:
            return 9999
        cost += graph[u][v]
    return cost
def generate_route():
    path = [source]
    current = source
    visited = {source}
    while current != destination:
        neighbors = [
            i for i in range(N)
            if graph[current][i] > 0 and i not in visited
        ]
        if not neighbors:
            return [source, destination]
        current = random.choice(neighbors)
        path.append(current)
        visited.add(current)
    return path
nests = [generate_route() for _ in range(N)]
best = min(nests, key=fitness)
for iteration in range(T):
    for i in range(N):
        new_route = generate_route()
        j = random.randrange(N)
        if fitness(new_route) < fitness(nests[j]):
            nests[j] = new_route
    for i in range(N):
        if random.random() < Pa:
            nests[i] = generate_route()
    current_best = min(nests, key=fitness)
    if fitness(current_best) < fitness(best):
        best = current_best
print("Source Router:", source)
print("Destination Router:", destination)
print("Best Data Packet Route:", " -> ".join(map(str, best)))
print("Minimum Routing Cost:", fitness(best))
```
